## Notes

- `Global_Sales` is the target variable for the regression task defined above.
- `NA_Sales`, `EU_Sales`, `JP_Sales`, and `Other_Sales` are excluded from the model's input features because they algebraically sum to `Global_Sales`, which would leak the target.
- Categorical columns (`Platform`, `Genre`, `Publisher`) will need encoding (e.g. one-hot encoding) before model training in a future lab.
